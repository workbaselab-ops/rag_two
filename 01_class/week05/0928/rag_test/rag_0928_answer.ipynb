# 가상Tech 업무 가이드 RAG - 정답

- 권장 Loader와 chunk 전략, ChromaDB 영구 저장, 검색 테스트 15문항, 정답 chunk 위치 및 평가 기준을 구현. 
- 동일한 PDF의 DB가 있으면 전체 문서를 다시 임베딩하지 않음.

````python
from dotenv import load_dotenv
load_dotenv()

import os
PDF_PATH = "./data/virtual_tech_internal_handbook.pdf"

if not os.path.exists(PDF_PATH):
    raise FileNotFoundError(f"PDF를 찾을 수 없습니다: {PDF_PATH}")
print(f"PDF 경로: {PDF_PATH}")
````

## 1. 문서 Loader

- `pypdf.PdfReader`로 PDF를 읽고 `langchain_core.documents.Document`로 변환.
- 페이지 메타데이터를 직접 관리. 
- 스캔 이미지 PDF라면 별도의 OCR 도구가 필요.(현재는 text로만 구성함)

````python
from pypdf import PdfReader
from langchain_core.documents import Document

# ---------- PDF 로더 및 Document 객체설정(metadata 기본설정) ----------
def load_pdf_by_page(pdf_path):
    """PDF를 페이지별로, LangChain Document로 변환 """

    # 문서 로딩
    loader = PdfReader(pdf_path)
    
    # documents 초기화
    documents = []

    for page_index, pdf_page in enumerate(loader.pages):
        documents.append(
            Document(
                # 텍스트 추출 예외처리(이미지는 ""로 처리) 
                page_content=pdf_page.extract_text() or "",
                # 메타데이터 설정(문서경로, 페이지index, 페이지번호)
                metadata={
                    "source": pdf_path,
                    "page": page_index,
                    "page_number": page_index + 1,
                },
            )
        )

    return documents

pages = load_pdf_by_page(PDF_PATH)


# ---------- Document 객체설정 추가(metadata 확장) ----------
section_by_page = {
    1: "표지 및 문서 정보", 2: "회사와 업무 원칙",
    3: "계정·기기·정보보안", 4: "업무 도구와 커뮤니케이션",
    5: "문서 작성과 지식 관리", 6: "개발·변경·배포",
    7: "장애 대응과 고객 커뮤니케이션", 8: "휴가·비용·복지",
    9: "FAQ와 연락처",
}
for page in pages:
    page_number = page.metadata["page_number"]
    # 메타데이터 수정.
    page.metadata.update(
        # 페이지 번호
        # page_number=page_number,
        # dict.get(key, default_value): 섹션값으로 번호를 할당하고, 번호가 없으면 "기타"로 할당.
        section=section_by_page.get(page_number, "기타"),
        source_name=os.path.basename(PDF_PATH),
    )
print(f"로드된 페이지: {len(pages)}")
````

## 2. chunk 전략

- 페이지 경계를 유지한 채 `RecursiveCharacterTextSplitter`로 분할. 
- `chunk_size=800`, `chunk_overlap=120`은 정책의 조건·예외·수치를 함께 보존 및 검색 범위를 좁힘. 
- 한국어 문단과 문장 종결 표현을 우선 구분자로 사용하고 `p페이지-c순번` 형식의 chunk ID를 부여.

````python
from collections import defaultdict

from langchain_text_splitters import RecursiveCharacterTextSplitter

# ---------- 문서 청킹 ----------
splitter = RecursiveCharacterTextSplitter(
    # 최대 청크길이
    chunk_size=800,
    # 문맥 중복 청크길이
    chunk_overlap=120,
    # 청킹 단위(문단>줄바꿈>마침표+공백>다.>요.>띄어쓰기>글자)순으로 나누기
    separators=["\n\n", "\n", ". ", "다.", "요.", " ", ""],
)
chunks = splitter.split_documents(pages)


# ---------- 페이지별 청크 카운팅 및 청크id 생성 ----------

# dict(int) : 키값이 없으면 오류남.
# defaultdict: 키값이 없으면 0으로 설정함.(예외처리 역할)
page_counters = defaultdict(int)
for chunk in chunks:
    page_number = chunk.metadata["page_number"]
    page_counters[page_number] += 1
    # 02d : 두자리 정리 포맷설정
    chunk.metadata["chunk_id"] = f"p{page_number:02d}-c{page_counters[page_number]:02d}"

print(f"생성된 chunk: {len(chunks)}")


# ---------- 출력결과 미리보기 ----------
for chunk in chunks:
    # replace("\n", " ") : 줄바꿈 문자를 공백으로 치환.
    preview = chunk.page_content.replace("\n", " ")[:70]

    # chunk_id 출력, section 출력, [:70] 앞 ~ 70자를 잘라서 출력.
    print(chunk.metadata["chunk_id"], chunk.metadata["section"], preview)
````

## 3. ChromaDB 영구 저장

```code
- Hash를 이용한 DB_PATH설정 목적은, 데이터 무결성 검증과 파일 기반 캐싱(Caching)을 위한 시스템적 안정성에 집중하기 위해.
- 문서 내용 변경시, 자동으로 변경을 감지하여 Hash값이 바뀜.
- 중복 되는 저장과 계산을 방지.
- 부수적인 목적으로, 정보 노출 최소화와 경로 탐색 공격 방지됨.
```
- PDF 해시Hash 값을 추출해서, DB 폴더명으로 사용. 
- PDF가 같으면 기존 DB를 불러오고, PDF가 바뀌면 새 DB를 생성.(같은 문서를 반복 임베딩하지 않기위해)

````python
from hashlib import sha256
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

# ---------- PDF Hash기반 DB 경로 설정 ----------

# "rb" : 읽기전용Binary이진데이터(0또는1)[.pdf, .png, .mp4, .zip 등]
with open(PDF_PATH, "rb") as pdf_file:
    # 이진수12자리생성
    pdf_hash = sha256(pdf_file.read()).hexdigest()[:12]
DB_PATH = f"./chroma_db/{pdf_hash}"

# ---------- 컬렉션 네임 설정 ----------
COLLECTION_NAME = "virtualtech-guide"

# ---------- 임베딩 모델 설정 ----------
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")


# ---------- 조건부 DB로드 및 캐싱(If-Else) ----------
if os.path.exists(f"{DB_PATH}/chroma.sqlite3"):
    vectorstore = Chroma(
        collection_name=COLLECTION_NAME,
        persist_directory=DB_PATH,
        embedding_function=embeddings,
    )
    print(f"기존 DB 로드: {DB_PATH}")
else:

    # "chroma.sqlite3" DB가 없는 경우에만 실행함.
    vectorstore = Chroma.from_documents(
        documents=chunks, 
        embedding=embeddings,
        collection_name=COLLECTION_NAME, 
        persist_directory=DB_PATH,
        # p01-c01 형태의 chunk_id를,  DB의 ids=[...]로 지정함.
        ids=[chunk.metadata["chunk_id"] for chunk in chunks],
    )
    print(f"새 DB 생성: {DB_PATH}")

# ---------- 리트리버 (유사도 상위4개 청크를 결과로 전달) ----------
retriever = vectorstore.as_retriever(search_kwargs={"k": 4})
````

## 5. 평가 기준 - 정량적 평가

### 검색 품질 50점
- Hit@4 30점: 상위 4개 결과에 정답 페이지가 포함된 비율
- MRR@4 15점: 정답 페이지가 높은 순위에 나올수록 높은 점수
- 근거 키워드 5점: 검색 문맥에 정답 anchor가 포함된 비율

### 답변 품질 50점
- 사실 정확성 20점: 수치·시간·승인자·금지사항 일치
- 근거 충실성 15점: 문맥에 없는 내용을 생성하지 않음
- 완전성 10점: 복수 조건과 예외를 빠뜨리지 않음
- 출처 표시 5점: 실제 근거 chunk ID 표시

``` 
권장 통과 기준은 `Hit@4 >= 0.90`, `MRR@4 >= 0.75`, 총점 80점 이상. 
보안·권한·장애 대응 정책을 반대로 답하거나 없는 정책을 생성하면 치명 오류임.
```

````python
# ---------- 검색 결과 도출 함수 ----------
def show_search_results(question, k=4):

    # 유사도 검색 실행 : 질문과 코사인 유사도 점수(score)가 높은 순으로 k개 문서(doc) 추출.
    results = vectorstore.similarity_search_with_relevance_scores(question, k=k)
    # 출력
    for rank, (doc, score) in enumerate(results, 1):
        preview = doc.page_content.replace("\n", " ")[:140]
        print(f"{rank}. {doc.metadata['chunk_id']} | p.{doc.metadata['page_number']} | {score:.3f} | {preview}")
    return results


# ---------- 검색 모델 성능 평가(3가지 핵심 평가 지표) 함수 ----------
def evaluate_retrieval(cases, k=4):
    details = []
    for case in cases:
        
        docs = vectorstore.similarity_search(case["question"], k=k)

        found_pages = [d.metadata["page_number"] for d in docs]
        hit = case["page"] in found_pages
        rank = found_pages.index(case["page"]) + 1 if hit else None
        context = " ".join(d.page_content.lower() for d in docs)
        anchor_hit = any(a.lower() in context for a in case["anchors"])
        details.append({"id": case["id"], "hit": hit, "rank": rank, "anchor_hit": anchor_hit})

    # 검색된 상위 k개 문서 안에 정답 페이지가 포함되었는가(성공 비율)
    hit_at_k = sum(d["hit"] for d in details) / len(details)

    # 정답 페이지가 상단 몇 번째에 나왔는가를 측정 (1위면 1점, 2위면 0.5점)
    mrr_at_k = sum(1 / d["rank"] if d["rank"] else 0 for d in details) / len(details)

    # 검색된 문서들의 본문에 핵심 키워드(anchors)가 최소 1개 이상 들어있는가
    anchor_rate = sum(d["anchor_hit"] for d in details) / len(details)

    # 소수3자기까지 출력
    print(f"Hit@{k}: {hit_at_k:.3f}")
    print(f"MRR@{k}: {mrr_at_k:.3f}")
    print(f"Anchor hit rate: {anchor_rate:.3f}")
    return {"hit_at_k": hit_at_k, "mrr_at_k": mrr_at_k, "anchor_rate": anchor_rate, "details": details}


# # ---------- RAG 수행결과 및 성능 테스트 ----------

# # 특정 질문(8번 케이스) 검색 결과 상세 보기
# show_search_results(test_cases[8]["question"])

# print("--------------------")

# # 전체 테스트 케이스에 대한 검색 성능 레포트 출력
# report = evaluate_retrieval(test_cases, k=4)
````

## 6. 근거 기반 답변 생성

- 검색 문맥만 사용하고 답변 끝에 실제 chunk ID를 표시. 
- 실행하면 LLM 토큰 비용이 발생.

````python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4.1-mini", temperature=0)

# 환각Hallucination방지를 위한 제약조건 부여.
SYSTEM_PROMPT = """
제공된 문맥만 사용해 한국어로 답하세요. 수치, 시간, 승인자와 예외를 빠뜨리지 마세요.
답이 없으면 '제공된 문서에서 확인할 수 없습니다.'라고 답하세요.
마지막에 사용한 chunk ID를 [근거: p02-c01] 형식으로 표시하세요.
""".strip()

def ask(question, k=4):
    # 유사도
    docs = vectorstore.similarity_search(question, k=k)

    # 문서청크의 메타데이터 바인딩 - 출처 추적함.
    context = "\n\n".join(
        f"[{d.metadata['chunk_id']}] p.{d.metadata['page_number']}\n{d.page_content}"
        for d in docs
    )

    # 리스트[(튜플),(튜플)] 
    response = llm.invoke([
        ("system", SYSTEM_PROMPT),
        ("human", f"문맥:\n{context}\n\n질문: {question}"),
    ])
    return response.content

# # 아래요청의 정답은, 하나뿐임.
# print(ask("고위험 배포는 누구의 승인을 받아 언제 진행하나요?"))
````
