```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_community.document_loaders import PyMuPDFLoader

FILE_PATH = "./data/virtual_tech_internal_handbook.pdf"

loader = PyMuPDFLoader(FILE_PATH)
docs = loader.load()

print(len(docs))
print(docs[0].page_content[:200])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100,
)

split_docs = text_splitter.split_documents(docs)

print(len(split_docs))
```

```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = OpenAIEmbeddings()
vectorstore = FAISS.from_documents(split_docs, embeddings)
```

```python
retriever = vectorstore.as_retriever(serch_kwargs={"k": 4})

found = retriever.invoke("연차를 사용하려면 최소 며칠 전에 신청해야 하나요?")
for i,d in enumerate(found, 1):
    print(i, d.metadata.get("page"), d.page_content[:60].replace("\n", " "))
```
